```python
import pandas as pd
import numpy as np
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun",
          "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
revenue_values  = [42, 45, 50, 48, 55, 60, 58, 62, 65, 70, 75, 85]
expense_values  = [38, 40, 42, 47, 50, 52, 55, 60, 61, 64, 66, 72]
```

```python

#A1. Create revenue and expenses Series using months as the index. Name them "Revenue" and "Expenses"
revenue=pd.Series(revenue_values,index=months,name='revenue')
expense=pd.Series(revenue_values,index=months,name='expense')
print(revenue)
print(expense)

```

```python
#A2.Let a = [42, 45, 50] and b = [38, 40, 42]. Display a + b and explain why it does not give the profit for each month.
a = [42, 45, 50]
b = [38, 40, 42]
profit=a+b
print(profit)
```

```python
#A3.Calculate profit for all 12 months using a for loop that appends to a list.
profit_list=[]
for i in range(12):
    profit_list.append(revenue_values[i] - expense_values[i])
print(profit_list)
```

```python
#A4.Calculate the same result with a single Series expression and confirm it matches A3.
profit=revenue - expense
print(profit)
```

```python
#B1.Create profit = revenue - expenses and name it "Profit".
profit_list=revenue-expense
profit.name='profit'
print(profit_list)
```

```python
#B2.Create margin, the profit margin %, as profit / revenue * 100, rounded to 1 decimal.
margin=round(profit/revenue * 100,1)
print(margin)
```

```python
#B3.Calculate total revenue, total expenses, total profit and the overall margin % for the year.
total_revenue=revenue.sum()
print(total_revenue)
```

```python
total_expense=expense.sum()
print(total_expense)
```

```python
#E1.Define profit_status(p): 8 or above = "Strong", 4 to below 8 = "Stable", below 4 = "Weak". Apply it to profit.
def profit_status(p):
    if p >= 8:
        return "Strong"
    elif p >= 4:
        return "Stable"
    else:
        return "Weak"

status = profit.apply(profit_status)
print(status)
```

```python
#E2.Display the number and percentage of months in each status.
count = status.value_counts()
percentage = (count / len(status) * 100).round(1)

print(count)
print(percentage)
```

```python
#E3.Define tax_payable(p): 25% of profit if profit is positive, otherwise 0. Apply it to profit.
def tax_payable(p):
    return p * 0.25 if p > 0 else 0

tax = profit.apply(tax_payable)
print(tax)
```

```python
#E4.Calculate total tax and the post-tax profit for the year.
print("Total Tax:", tax.sum())
print("Post-Tax Profit:", profit.sum() - tax.sum())
```
